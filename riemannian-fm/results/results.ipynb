### Specify Results
- Change the experiment variable to applicable experiment
- change the x or y axis to get metrics

````python
def print_metrics(experiment: pd.DataFrame, x: str, y: str):
    print(f"Available metrics:")
    print(", ".join(map(str, experiment.columns.tolist())))
    print()
    
    selected = experiment[[experiment.columns[0], x, y]]
    print(selected.to_string(index=False))
````

````python
# specify metrics to compare
x = 'test_sample/volume_scaling_gap_target_minus_pred'
y = 'test_sample/sinkhorn_knopp'

print_metrics(metrics_gtg, x=x, y=y)
print_metrics(metrics_gtg_norm, x=x, y=y)
print_metrics(metrics_mog, x=x, y=y)
print_metrics(metrics_gaussian_ring, x=x, y=y)
print_metrics(metrics_checkerboard, x=x, y=y)
````
